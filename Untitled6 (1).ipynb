```python
# Calculate NDVI

def add_ndvi(image):

    ndvi = image.normalizedDifference(["B8", "B4"]).rename("NDVI")

    return image.addBands(ndvi)

sentinel2 = sentinel2.map(add_ndvi)
```

```python
# Calculate EVI

def add_evi(image):

    evi = image.expression(
        "2.5 * ((NIR - RED) / (NIR + 6 * RED - 7.5 * BLUE + 1))",
        {
            "NIR": image.select("B8"),
            "RED": image.select("B4"),
            "BLUE": image.select("B2"),
        },
    ).rename("EVI")

    return image.addBands(evi)

sentinel2 = sentinel2.map(add_evi)
```

```python
# Calculate NDWI

def add_ndwi(image):

    ndwi = image.normalizedDifference(["B8", "B11"]).rename("NDWI")

    return image.addBands(ndwi)

sentinel2 = sentinel2.map(add_ndwi)
```

```python
# Scale optical and thermal bands

def scale_landsat(image):

    optical = image.select("SR_B.*").multiply(0.0000275).add(-0.2)

    thermal = image.select("ST_B10").multiply(0.00341802).add(149.0)

    return (
        image
        .addBands(optical, overwrite=True)
        .addBands(thermal, overwrite=True)
    )

landsat8 = landsat8.map(scale_landsat)
```

```python
# Keep only Land Surface Temperature

def select_lst(image):

    return image.select("ST_B10").rename("LST")

landsat8 = landsat8.map(select_lst)
```

```python
# Number of Sentinel-1 images per month

months = ee.List.sequence(1, 12)

def monthly_count(month):
    count = sentinel1.filter(ee.Filter.calendarRange(month, month, "month")).size()
    return ee.Feature(None, {
        "Month": month,
        "Images": count
    })

monthly_s1 = ee.FeatureCollection(months.map(monthly_count))

print(monthly_s1.getInfo())
```

```python
# Number of Sentinel-2 images per month

def monthly_count(month):
    count = sentinel2.filter(ee.Filter.calendarRange(month, month, "month")).size()
    return ee.Feature(None, {
        "Month": month,
        "Images": count
    })

monthly_s2 = ee.FeatureCollection(months.map(monthly_count))

print(monthly_s2.getInfo())
```

```python
# Number of Landsat 8 images per month

def monthly_count(month):
    count = landsat8.filter(ee.Filter.calendarRange(month, month, "month")).size()
    return ee.Feature(None, {
        "Month": month,
        "Images": count
    })

monthly_l8 = ee.FeatureCollection(months.map(monthly_count))

print(monthly_l8.getInfo())
```

```python
# Create monthly Sentinel-1 composites

years = ee.List.sequence(2017, 2025)
months = ee.List.sequence(1, 12)

def create_monthly_s1(year):

    year = ee.Number(year)

    def monthly_image(month):

        month = ee.Number(month)

        start = ee.Date.fromYMD(year, month, 1)
        end = start.advance(1, "month")

        image = (
            sentinel1
            .filterDate(start, end)
            .median()
            .set("year", year)
            .set("month", month)
            .set("system:time_start", start.millis())
        )

        return image

    return months.map(monthly_image)

sentinel1_monthly = ee.ImageCollection(
    years.map(create_monthly_s1).flatten()
)
```

```python
print("Sentinel-1 Monthly Images:")

print(sentinel1_monthly.size().getInfo())
```

```python
# Create monthly Sentinel-2 composites

years = ee.List.sequence(2017, 2025)
months = ee.List.sequence(1, 12)

def create_monthly_s2(year):

    year = ee.Number(year)

    def monthly_image(month):

        month = ee.Number(month)

        start = ee.Date.fromYMD(year, month, 1)
        end = start.advance(1, "month")

        return (
            sentinel2
            .filterDate(start, end)
            .median()
            .set("year", year)
            .set("month", month)
            .set("system:time_start", start.millis())
        )

    return months.map(monthly_image)

sentinel2_monthly = ee.ImageCollection(
    years.map(create_monthly_s2).flatten()
)
```

```python
# Create monthly Landsat-8 composites

years = ee.List.sequence(2017, 2025)
months = ee.List.sequence(1, 12)

def create_monthly_l8(year):

    year = ee.Number(year)

    def monthly_image(month):

        month = ee.Number(month)

        start = ee.Date.fromYMD(year, month, 1)
        end = start.advance(1, "month")

        image = (
            landsat8
            .filterDate(start, end)
            .median()
            .set("year", year)
            .set("month", month)
            .set("system:time_start", start.millis())
        )

        return image

    return months.map(monthly_image)

landsat8_monthly = ee.ImageCollection(
    years.map(create_monthly_l8).flatten()
)

print("Landsat-8 Monthly Images:")
print(landsat8_monthly.size().getInfo())
```

```python
# --- دمج الأقمار الثلاثة وضغطها إلى Int16 ---

def stack_and_compress(s2_img):
    s2_img = ee.Image(s2_img)
    time_start = s2_img.get('system:time_start')

    # سحب صور S1 و L8 المطابقة لنفس الشهر
    s1_img = ee.Image(sentinel1_monthly.filter(ee.Filter.eq('system:time_start', time_start)).first())
    l8_img = ee.Image(landsat8_monthly.filter(ee.Filter.eq('system:time_start', time_start)).first())

    # اختيار القنوات الـ 6 الأساسية فقط
    s1_bands = s1_img.select(['VV', 'VH'])
    s2_bands = s2_img.select(['NDVI', 'EVI', 'NDWI'])
    l8_bands = l8_img.select(['LST'])

    # الدمج في مصفوفة واحدة (Stack)
    stacked = s2_bands.addBands(s1_bands).addBands(l8_bands)

    # الضغط: الضرب بـ 10000 والتحويل إلى عدد صحيح (Int16) لتوفير المساحة
    compressed = stacked.multiply(10000).toInt16()

    return compressed.set({
        'system:time_start': time_start,
        'year': s2_img.get('year'),
        'month': s2_img.get('month')
    })

# بناء المجموعة النهائية
final_stacked_collection = ee.ImageCollection(sentinel2_monthly.map(stack_and_compress))

print("تم بناء عدد الصور الكلي للSTACK")
```

```python
# --- حساب الحجم التقديري للبيانات قبل التصدير ---

# 1. ثوابت المشروع
scale = 10              # الدقة المكانية (10 أمتار)
num_bands = 6           # عدد القنوات المدمجة (VV, VH, NDVI, EVI, NDWI, LST)
bytes_per_pixel = 2     # حجم التخزين (لأننا استخدمنا toInt16 لضغطها)

# 2. سحب المربع الأول من شبكة الأحساء المفلترة
first_grid = grid_ee.first()

# 3. حساب مساحة المربع الجغرافي بالمتر المربع
grid_area_sq_meters = first_grid.geometry().area().getInfo()

# 4. العمليات الحسابية
pixel_area = scale * scale  # مساحة البكسل الواحد (100 متر مربع)
total_pixels = grid_area_sq_meters / pixel_area  # إجمالي البكسلات
total_bytes = total_pixels * num_bands * bytes_per_pixel  # الحجم بالبايت

# التحويل إلى ميجابايت (MB) وجيجابايت (GB)
size_mb = total_bytes / (1024 * 1024)
size_gb_all_months = (size_mb * 108) / 1024

print(" --- تقرير المساحة التقديرية ---")
print(f" مساحة المربع الفعلي: {grid_area_sq_meters / 1e6:.2f} كيلومتر مربع")
print(f" الحجم الخام للصورة الواحدة (للمربع في شهر واحد): {size_mb:.2f} ميجابايت")
print(f" الحجم الإجمالي لهذا المربع عبر كل السنوات (108 شهر): {size_gb_all_months:.2f} جيجابايت")
```

```python
# أخذ المربع الأول من الشبكة
first_grid = grid_ee.first()
grid_id = first_grid.get('grid_id').getInfo()
grid_geom = first_grid.geometry()

# أخذ الإسقاط الأصلي لتجنب مشاكل التطابق (نستدعيها مرة واحدة فقط)
s2_crs = first_s2.select('B4').projection().getInfo()['crs']

# القوائم الخاصة بالسنين والشهور في بايثون
years = list(range(2017, 2026))
months = list(range(1, 13))

print(f"بدء إرسال مهام التصدير للمربع: {grid_id} ...")

# نلف على السنوات والشهور من بايثون مباشرة لتوفير الذاكرة
for year in years:
    for month in months:

        # 1. استخراج الصورة من المجموعة بناءً على السنة والشهر
        img = ee.Image(
            final_stacked_collection.filter(
                ee.Filter.And(
                    ee.Filter.eq('year', year),
                    ee.Filter.eq('month', month)
                )
            ).first()
        )

        # 2. بناء اسم الملف من متغيرات بايثون (تخلصنا من getInfo)
        file_name = f"{grid_id}_{year}_{month:02d}"

        # 3. إرسال المهمة لسيرفرات جوجل الخلفية براحة تامة
        task = ee.batch.Export.image.toDrive(
            image=img,
            description=file_name,
            folder=f'Drought_Grid_{grid_id}',
            fileNamePrefix=file_name,
            region=grid_geom,
            scale=10,
            crs=s2_crs,
            maxPixels=1e13
        )
        task.start()
```

```python
# --- كود تصفية المربعات الصحراوية (Desert Grid Filtering) ---

print(" جاري فحص خريطة الأحساء واستبعاد مربعات الصحراء القاحلة (هذه العملية قد تستغرق دقيقة)...")

# 1. جلب خريطة الغطاء الأرضي من وكالة الفضاء الأوروبية
landcover = ee.ImageCollection("ESA/WorldCover/v200").first()

# في تصنيف وكالة الفضاء، رقم 60 يمثل (الصحراء والأراضي القاحلة Barren)
# الكود التالي يصنع قناع (Mask) يعطي قيمة 1 لأي شيء غير الصحراء، و0 للصحراء
non_desert_mask = landcover.neq(60)

# 2. دالة تفحص كل مربع: تحسب كم بكسل "غير صحراوي" موجود داخله
def filter_desert(feature):
    stats = non_desert_mask.reduceRegion(
        reducer=ee.Reducer.sum(),
        geometry=feature.geometry(),
        scale=100, # استخدمنا مقياس 100 متر لتسريع الحسبة على السيرفر
        maxPixels=1e9
    )
    # حفظ النتيجة داخل المربع
    return feature.set('useful_pixels', stats.get('Map'))

# 3. تطبيق الفحص على شبكة الـ 1008 مربعات
grids_with_stats = grid_ee.map(filter_desert)

# 4. التصفية النهائية: الاحتفاظ فقط بالمربعات اللي تحتوي على مناطق غير صحراوية
useful_grids_ee = grids_with_stats.filter(ee.Filter.gt('useful_pixels', 50))

# تحديث المتغير القديم ليحتوي على المربعات المفيدة فقط
grid_ee = useful_grids_ee

# طباعة العدد النهائي بعد التصفية
final_count = grid_ee.size().getInfo()
print("="*50)
print(f"تمت التصفية بنجاح!")
print(f" العدد الإجمالي القديم: 1008 مربع")
print(f" العدد الجديد المفيد (بدون صحراء): {final_count} مربع فقط")
print("="*50)
```

```python
# --- كود الاستكمال الذكي (يرسل المربعات المفيدة المتبقية فقط) ---

# تعريف المتغيرات الأساسية لتجنب أخطاء الذاكرة
time_series_image = final_stacked_collection.toBands()
s2_crs = ee.ImageCollection("COPERNICUS/S2_SR_HARMONIZED").first().select('B4').projection().getInfo()['crs']

# 1. جلب أسماء المربعات المفيدة (المتبقية بعد فلتر الصحراء)
useful_grid_ids = grid_ee.aggregate_array('grid_id').getInfo()

# 2. فحص طابور المهام لمعرفة المربعات اللي أرسلتها (الـ 171 مهمة)
tasks = ee.batch.Task.list()
# نستخرج اسم المربع (مثل AHN-00001) من المهام المرسلة
sent_grid_ids = [task.config.get('description', '').split('_Time_Series')[0]
                 for task in tasks if 'Time_Series' in task.config.get('description', '')]

# 3. استخراج المربعات اللي ه ما انرسلت (المقاطعة بين القائمتين)
remaining_grids = [gid for gid in useful_grid_ids if gid not in sent_grid_ids]

print(f" إجمالي المربعات المفيدة (بدون صحراء): {len(useful_grid_ids)}")
print(f" المربعات المفيدة اللي انرسلت ضمن الـ 171: {len(useful_grid_ids) - len(remaining_grids)}")
print(f" المربعات المفيدة المتبقية وبنرسلها الآن: {len(remaining_grids)}\n" + "="*50)

# 4. إرسال المتبقي فقط
if len(remaining_grids) > 0:
    for current_id in remaining_grids:
        # جلب هندسة المربع الحالي على الطاير
        current_feature = grid_ee.filter(ee.Filter.eq('grid_id', current_id)).first()
        grid_geom = current_feature.geometry()

        file_name = f"{current_id}_Time_Series_2017_to_2025"

        task = ee.batch.Export.image.toDrive(
            image=time_series_image,
            description=file_name,
            folder='Drought_Project_Optimized',
            fileNamePrefix=file_name,
            region=grid_geom,
            scale=10,
            crs=s2_crs,
            maxPixels=1e13
        )
        task.start()
        print(f" تم إرسال المهمة للمربع المتبقي: {current_id}")

    print("="*50)
    print(" تم إرسال جميع المربعات المفيدة بنجاح! طابور المهام الآن مكتمل للمشروع.")
else:
    print(" مبروك! كل المربعات المفيدة تم إرسالها مسبقاً وما يحتاج ترسل شيء جديد.")
```

```python
# --- كود لوحة المعلومات لمتابعة حالة الـ 170 مربع المفيدة ---

tasks = ee.batch.Task.list()

# قواميس لتتبع عدد الحالات
status_counts = {'READY': 0, 'RUNNING': 0, 'COMPLETED': 0, 'FAILED': 0, 'CANCELLED': 0}
useful_tasks = []

print("📊 جاري جمع بيانات المهام المفيدة (أكبر من 171)...\n")

# فلترة المهام
for task in tasks:
    desc = task.config.get('description', '')

    # التأكد أن المهمة تابعة لمشروعنا
    if 'Time_Series' in desc and desc.startswith('AHN-'):
        try:
            grid_num = int(desc.split('_')[0].replace('AHN-', ''))

            # نركز فقط على المربعات المفيدة اللي أرسلناها (رقمها فوق 171)
            if grid_num > 171:
                useful_tasks.append(task)
                state = task.state

                if state in status_counts:
                    status_counts[state] += 1
        except Exception:
            pass

# طباعة الملخص الإحصائي
print("="*45)
print(" ملخص حالة المربعات المفيدة (170 مربع):")
print("="*45)
print(f" في طابور الانتظار (READY)    : {status_counts['READY']}")
print(f" قيد المعالجة الآن (RUNNING) : {status_counts['RUNNING']}")
print(f" اكتملت بنجاح (COMPLETED)   : {status_counts['COMPLETED']}")
print(f" فشلت (FAILED)              : {status_counts['FAILED']}")
print("="*45)

# عرض عينة للمهام النشطة
print("\n عينة من الملفات اللي شغالة أو تنتظر حالياً:")
active_count = 0
for task in useful_tasks:
    if task.state in ['READY', 'RUNNING']:
        print(f"الملف: {task.config.get('description', '')} ---> الحالة: {task.state}")
        active_count += 1
        if active_count == 5:  # يعرض أول 5 فقط عشان الترتيب
            break


```
