```python
boundary = gpd.read_file("al_ahsa_boundary (4).geojson")
grid = gpd.read_file("al_ahsa_grid (4).geojson")
```

```python
boundary_ee = ee.FeatureCollection(boundary.__geo_interface__)
grid_ee = ee.FeatureCollection(grid.__geo_interface__)
```

```python
Map = geemap.Map()

Map.centerObject(boundary_ee, 8)

Map.addLayer(boundary_ee, {"color": "red"}, "Boundary")

Map
```

```python
start_date = "2017-01-01"
end_date = "2025-12-31"
```

```python
sentinel1 = (
    ee.ImageCollection("COPERNICUS/S1_GRD")
    .filterBounds(boundary_ee)
    .filterDate(start_date, end_date)
    .filter(ee.Filter.eq("instrumentMode", "IW"))
    .filter(ee.Filter.listContains("transmitterReceiverPolarisation", "VV"))
    .filter(ee.Filter.listContains("transmitterReceiverPolarisation", "VH"))
)
```

```python
first_s1 = sentinel1.first()
```

```python
# Clip every Sentinel-1 image to the Al-Ahsa boundary

def clip_image(image):

    return image.clip(boundary_ee)

sentinel1 = sentinel1.map(clip_image)
```

```python
print("Number of Sentinel-1 Images:")

print(sentinel1.size().getInfo())
```

```python
first_s1 = sentinel1.first()

print(first_s1.bandNames().getInfo())
```

```python
# Keep only the VV and VH bands

def select_bands(image):

    return image.select(["VV", "VH"])

sentinel1 = sentinel1.map(select_bands)
```

```python
first_s1 = sentinel1.first()

print(first_s1.bandNames().getInfo())
```

```python
Map = geemap.Map()

Map.centerObject(boundary_ee, 8)

Map.addLayer(
    first_s1.select("VV"),
    {
        "min": -25,
        "max": 5
    },
    "VV"
)

Map.addLayer(
    first_s1.select("VH"),
    {
        "min": -30,
        "max": 0
    },
    "VH"
)

Map.addLayer(
    boundary_ee,
    {
        "color": "red"
    },
    "Boundary"
)

Map
```

```python
sentinel2 = (
    ee.ImageCollection("COPERNICUS/S2_SR_HARMONIZED")
    .filterBounds(boundary_ee)
    .filterDate("2017-01-01", "2025-12-31")
)
```

```python
print("Total Sentinel-2 Images:")

print(sentinel2.size().getInfo())
```

```python
first_s2 = sentinel2.first()

print(first_s2.bandNames().getInfo())
```

```python
# Remove cloudy images

sentinel2 = sentinel2.filter(
    ee.Filter.lt("CLOUDY_PIXEL_PERCENTAGE", 20)
)
```

```python
print("Images after cloud filtering:")

print(sentinel2.size().getInfo())
```

```python
# Clip every image to the Al-Ahsa boundary

def clip_image(image):

    return image.clip(boundary_ee)

sentinel2 = sentinel2.map(clip_image)
```

```python
first_s2 = sentinel2.first()

Map = geemap.Map()

Map.centerObject(boundary_ee, 8)

Map.addLayer(
    first_s2,
    {
        "bands": ["B4", "B3", "B2"],
        "min": 0,
        "max": 3000
    },
    "RGB"
)

Map.addLayer(
    boundary_ee,
    {
        "color": "red"
    },
    "Boundary"
)

Map
```

```python
# Calculate NDVI

def add_ndvi(image):

    ndvi = image.normalizedDifference(["B8", "B4"]).rename("NDVI")

    return image.addBands(ndvi)

sentinel2 = sentinel2.map(add_ndvi)
```

```python
# Calculate EVI

def add_evi(image):

    evi = image.expression(
        "2.5 * ((NIR - RED) / (NIR + 6 * RED - 7.5 * BLUE + 1))",
        {
            "NIR": image.select("B8"),
            "RED": image.select("B4"),
            "BLUE": image.select("B2"),
        },
    ).rename("EVI")

    return image.addBands(evi)

sentinel2 = sentinel2.map(add_evi)
```

```python
# Calculate NDWI

def add_ndwi(image):

    ndwi = image.normalizedDifference(["B8", "B11"]).rename("NDWI")

    return image.addBands(ndwi)

sentinel2 = sentinel2.map(add_ndwi)
```

```python
first_s2 = sentinel2.first()

print(first_s2.bandNames().getInfo())
```

```python
Map = geemap.Map()

Map.centerObject(boundary_ee, 8)

Map.addLayer(
    first_s2.select("NDVI"),
    {
        "min": -1,
        "max": 1,
        "palette": ["brown", "yellow", "green"],
    },
    "NDVI",
)

Map.addLayer(
    first_s2.select("EVI"),
    {
        "min": -1,
        "max": 1,
        "palette": ["brown", "yellow", "green"],
    },
    "EVI",
)

Map.addLayer(
    first_s2.select("NDWI"),
    {
        "min": -1,
        "max": 1,
        "palette": ["brown", "yellow", "blue"],
    },
    "NDWI",
)

Map.addLayer(
    boundary_ee,
    {
        "color": "red",
    },
    "Boundary",
)

Map
```

```python
landsat8 = (
    ee.ImageCollection("LANDSAT/LC08/C02/T1_L2")
    .filterBounds(boundary_ee)
    .filterDate("2017-01-01", "2025-12-31")
)
```

```python
print("Total Landsat 8 Images:")

print(landsat8.size().getInfo())
```

```python
first_l8 = landsat8.first()

print(first_l8.bandNames().getInfo())
```

```python
# Clip every Landsat 8 image to the Al-Ahsa boundary

def clip_image(image):

    return image.clip(boundary_ee)

landsat8 = landsat8.map(clip_image)
```

```python
# Remove cloudy images

landsat8 = landsat8.filter(
    ee.Filter.lt("CLOUD_COVER", 20)
)
```

```python
print("Images after cloud filtering:")

print(landsat8.size().getInfo())
```

```python
# Scale optical and thermal bands

def scale_landsat(image):

    optical = image.select("SR_B.*").multiply(0.0000275).add(-0.2)

    thermal = image.select("ST_B10").multiply(0.00341802).add(149.0)

    return (
        image
        .addBands(optical, overwrite=True)
        .addBands(thermal, overwrite=True)
    )

landsat8 = landsat8.map(scale_landsat)
```

```python
# Keep only Land Surface Temperature

def select_lst(image):

    return image.select("ST_B10").rename("LST")

landsat8 = landsat8.map(select_lst)
```

```python
first_l8 = landsat8.first()

print(first_l8.bandNames().getInfo())
```

```python
Map = geemap.Map()

Map.centerObject(boundary_ee, 8)

Map.addLayer(
    first_l8.select("LST"),
    {
        "min": 290,
        "max": 330,
        "palette": [
            "blue",
            "cyan",
            "green",
            "yellow",
            "orange",
            "red"
        ],
    },
    "LST",
)

Map.addLayer(
    boundary_ee,
    {
        "color": "red",
    },
    "Boundary",
)

Map
```

```python
# Number of Sentinel-1 images per month

months = ee.List.sequence(1, 12)

def monthly_count(month):
    count = sentinel1.filter(ee.Filter.calendarRange(month, month, "month")).size()
    return ee.Feature(None, {
        "Month": month,
        "Images": count
    })

monthly_s1 = ee.FeatureCollection(months.map(monthly_count))

print(monthly_s1.getInfo())
```

```python
# Number of Sentinel-2 images per month

def monthly_count(month):
    count = sentinel2.filter(ee.Filter.calendarRange(month, month, "month")).size()
    return ee.Feature(None, {
        "Month": month,
        "Images": count
    })

monthly_s2 = ee.FeatureCollection(months.map(monthly_count))

print(monthly_s2.getInfo())
```

```python
# Number of Landsat 8 images per month

def monthly_count(month):
    count = landsat8.filter(ee.Filter.calendarRange(month, month, "month")).size()
    return ee.Feature(None, {
        "Month": month,
        "Images": count
    })

monthly_l8 = ee.FeatureCollection(months.map(monthly_count))

print(monthly_l8.getInfo())
```

```python
# Create monthly Sentinel-1 composites

years = ee.List.sequence(2017, 2025)
months = ee.List.sequence(1, 12)

def create_monthly_s1(year):

    year = ee.Number(year)

    def monthly_image(month):

        month = ee.Number(month)

        start = ee.Date.fromYMD(year, month, 1)
        end = start.advance(1, "month")

        image = (
            sentinel1
            .filterDate(start, end)
            .median()
            .set("year", year)
            .set("month", month)
            .set("system:time_start", start.millis())
        )

        return image

    return months.map(monthly_image)

sentinel1_monthly = ee.ImageCollection(
    years.map(create_monthly_s1).flatten()
)
```

```python
print("Sentinel-1 Monthly Images:")

print(sentinel1_monthly.size().getInfo())
```

```python
# Create monthly Sentinel-2 composites

years = ee.List.sequence(2017, 2025)
months = ee.List.sequence(1, 12)

def create_monthly_s2(year):

    year = ee.Number(year)

    def monthly_image(month):

        month = ee.Number(month)

        start = ee.Date.fromYMD(year, month, 1)
        end = start.advance(1, "month")

        return (
            sentinel2
            .filterDate(start, end)
            .median()
            .set("year", year)
            .set("month", month)
            .set("system:time_start", start.millis())
        )

    return months.map(monthly_image)

sentinel2_monthly = ee.ImageCollection(
    years.map(create_monthly_s2).flatten()
)
```

```python
years = list(range(2017, 2026))

sentinel2_yearly = []

for year in years:

    months = ee.List.sequence(1, 12)

    def monthly_image(month):

        month = ee.Number(month)

        start = ee.Date.fromYMD(year, month, 1)
        end = start.advance(1, "month")

        return (
            sentinel2
            .filterDate(start, end)
            .median()
            .set("year", year)
            .set("month", month)
            .set("system:time_start", start.millis())
        )

    yearly = ee.ImageCollection.fromImages(
        months.map(monthly_image)
    )

    print(f"{year} :", yearly.size().getInfo())

    sentinel2_yearly.append(yearly)
```
